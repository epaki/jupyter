# LeachIT equation overview

## Gold Mass Fractions by Location in Particles

These equations compute the relative mass fractions of gold in different regions of a gold-bearing particle, depending on particle geometry and gold grain distribution. These are initial conditions for the leaching model.

### **Gold Near the external surface**
### $P_{AuL} = {{d_g}^3 + 2.25{d_g}^2(d_p - d_g) \over {d_p}^3}$

* **Purpose:** Estimates the mass fraction of gold close to the surface
* **Relevance:** Quickly dissolves during early leaching
* **Context:** $d_g$ is the gold grain size, $d_p$ is particle size.

### **Gold in the subsurface region**
### ​$P_{AuS} = {3{d_p}(d_p - d_g)^2 + 0.75{d_g}^2(d_p - d_g) \over {d_p}^3}$
 
* **Purpose:** Describes gold embedded in the outer shell but not at the surface.
* **Use:** Has a slower leaching rate than surface gold.

### **Gold in the middle zone**
### ​$P_{AuP} = {(d_p - d_g)^3 - (d_p - d_g - 2𝜎_p)^3 \over {d_p}^3}$
 
* **Purpose:** Fraction of gold in the central penetrable region.
* $σ_p$: Gold penetration depth by solution.

### **Unleachable gold in the core**
### $P_{Au∞} = {(d_p - d_g - 2𝜎_p)^3 \over {d_p}^3}$
 
* **Purpose:** Gold that cannot be leached under normal conditions.



## Leaching Rate and Mass Dissolution Equations
These equations describe the kinetics of gold dissolution from different particle zones using a pseudo-homogeneous rate law.

### **Gold dissolution rate (general form)**
### $v_{Au} = {4RD_{CN}[CN][O_2] \over σ_0(D_{CN}[CN] + 8RD_{O_2}[O_2])}$
* **Key variables:**
	* $𝐷_𝐶𝑁$, $𝐷_{𝑂_2}$: diffusivities of CN⁻ and O₂
	* $𝑅$: universal gas constant
	* $𝜎_0$: tortuosity or effectiveness factor
* **Purpose:** Calculates the rate at which gold dissolves into solution from the particle matrix.

### **The universal gas constant 𝑅**
R is a physical constant that appears in many fundamental equations in thermodynamics, chemistry, and physics, including the ideal gas law:

### $𝑃𝑉=𝑛𝑅𝑇$

Where:
* 𝑃: Pressure
* 𝑉: Volume
* 𝑛: Number of moles
* 𝑇: Temperature in Kelvin
* 𝑅: Universal gas constant

#### Value of the Universal Gas Constant
| Units           | Value of $R$ |
| --------------- | ------------ |
| J·mol⁻¹·K⁻¹     | 8.3145       |
| L·atm·mol⁻¹·K⁻¹ | 0.08206      |
| cal·mol⁻¹·K⁻¹   | 1.9872       |

In the De Andrade Lima & Hodouin paper, if 𝑅 is used in the kinetic expressions involving diffusion or surface velocity, it is most likely:
$R=8.314 J\cdotp mol {−1}\cdotp K^{−1}$

#### How It's Derived (Conceptually)
The gas constant 𝑅 is derived from Boltzmann’s constant $𝑘_𝐵$ multiplied by Avogadro’s number $𝑁_𝐴$:

$𝑅 = 𝑘_𝐵⋅𝑁_𝐴$
​ 
Where:
* $𝑘_𝐵 = 1.380649 × 10^{−23} J\cdotp K^{−1}$*(Boltzmann’s constant)*
* $𝑁_𝐴 = 6.02214076 × 10^{23}mol^{−1}$
	* $𝑅 = (1.380649 × 10^{−23}) \cdotp (6.02214076 × 10{23}) = 8.314462618 J \cdotp mol^{−1}⋅K^{−1}$


In the leaching model 𝑅 often appears in diffusion-based or reaction-rate equations that are adapted from gas or fluid transport theories—like the Nernst–Brunner or shrinking core model frameworks. Its inclusion helps convert energy terms (activation energy, etc.) into usable units that match the rest of the kinetic equation.

### **Mass change rate – External gold ($Au_E$)**
### ${dm_{AuE}(t) \over dt} = {M_{Au}m_{AuE}(0)^{1/3} \over 𝜌_{Au}d_g(0)} v_{AuE}m_{AuE}(t)^{2/3}$
* **Interpretation:** Rate of gold mass loss from external layer.
* **Note:** ⅓ and ⅔ exponents reflect shrinking particle model assumptions.

### **Mass change – Surface layer ($Au_L$)**
### ${dm_{AuL}(t) \over dt} = {6M_{Au}m_{AuL}(0)^{1/3} \over 𝜌_{Au}d_g(0)} v_{Au}m_{AuL}(t)^{2/3}$

* Higher coefficient (6 vs 1) due to geometry of exposed surface.
* Captures fast early-stage dissolution from outer shell.

### **Mass change – Penetrable core ($Au_p$)**
### ${dm_{Aup}(t) \over dt} = {M_{Au}m_{Aup}(0)^{1/3} \over 𝜌_{Au}d_g(0)} v_{Aup}m_{Aup}(t)^{2/3}$
* Similar structure to surface zone equation but for inner leachable core.


### **Mass change – Liquid phase ($Au_{Liq}$)**
### ${dm_{AuLiq}(t) \over dt} = {dm_{AuL}(t) \over dt} + {dm_{AuE}(t) \over dt} + {dm_{Aup}(t) \over dt}$
* **Purpose:** Aggregates all gold contributions to the liquid phase.

### **Modified Surface Velocity for External Gold**
### $v_{AuE}={4RD_{CN}[CN(t)]D_{O_2}[O_2(t)] \over (𝜎_0 + d_g(0)(1-{m_{AuE}(t) \over m_{AuE}(0)}))(D_{CN}[CN(t)] + 8RD_{O_2}[O_2(t)])}$



* Refined version of $ν_{Au}$ for external gold.
* Incorporates gold depletion over time: the term $1 - {m_{AuE}(t) \over m_{AuE}(0)}$ adjusts the reactivity as surface gold is leached.

## 🔍 How This Supports Cyanide Consumption & Dosage Modelling
These equations support dynamic simulation of gold leaching and can be used to:

#### Estimate cyanide requirements:**
* Dissolution rates depend on [CN⁻], [O₂], and particle size.
* Cyanide consumption also correlates with exposed gold surface and associated minerals (especially Cu, Fe).

#### Simulate leach circuit performance:
* Predict gold recovery vs time for different grind sizes.
* Track gold mass in each leachable region (surface, core, locked).

#### Optimise dosage:
* Too fine a grind increases cyanide demand (due to sulfide liberation).
* These equations help find a grind size that maximizes gold recovery while minimizing reagent cost.

#### Calibrate a plant-level model:
* Use these equations as the kinetic engine inside a plant simulation (e.g. in LeachOPTI).
* Adjust for site-specific constants, particle distributions, and measured CN losses.

### Kinetics
From Eq. (15) and (18) in the paper:
#### Gold dissolution rate:
$$r_{Au} = k_{Au} ([Au]_s - [Au]_{s,l})^{2.13} [CN^-]^{0.961} [O_2]^{0.228}$$

Where:
$$k_{Au} = 1.13 \times 10^{-3} - 4.37 \times 10^{-11} \cdot d^{2.93}$$
 
#### Cyanide consumption rate:
$$r_{CN^-} = \left( \frac{1.69 \times 10^{-8}}{d^{0.547}} - 6.40 \right) [CN^-]^{3.71}$$
