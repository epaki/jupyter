## ✅ **UI Use Case Summary**
#### **We want the user to:**
* See a 2D plan view of the block model
* Select/design initial pit development boundaries
* Provide directional hints (e.g. ramp direction, trim settings)
* Do this during development, not just after it’s all working

#### **So the frontend needs to be:**
* Interactive (mouse clicks, brush selection, sliders, dropdowns)
* Lightweight enough for fast iteration
* Future-proof for web deployment, but flexible for local use now

### 🧠 **Start With React + Vite + Plotly.js**

| **Option** | **Verdict** | **Why** |
| :--- | :--- | :--- |
| React	|✅ Best for future-proofing	| We already plan to go web-based. React supports component reuse, UI state, and full interactivity. |
| QT (e.g. PyQt/PySide) | 🟡 Ok for local-only tools | Slower to iterate, harder to share, not ideal for team use or cloud hosting. |
| Streamlit/Gradio | ⚠️ Great for quick forms, not suited to 2D spatial work | Can’t handle advanced drawing/plan views easily. |
| Jupyter Notebook UI | 🔧 Good for testing functions, not for interaction | You already outgrew this – not suitable for defining pit shells by clicking! |

### 🧪 **Hybrid workflow while developing and testing:**

#### 🧩 **Short Term:**
* Spin up a minimal React UI using Vite (very fast build tool)
* Load a static 2D projection of your block model (CSV or JSON)
* Let the user draw/select polygon(s) using react-plotly.js or react-leaflet
* Save their input as JSON (buffer shape, ramp hints, etc.)
* Backend reads this JSON and proceeds with logic

#### 🛠 **Tools we will use:**
|Tool|Purpose|
|:---|:---|
|`React + Vite`|Frontend framework|
|`react-plotly.js`|For 2D plan view plotting (uses Plotly under the hood)|
|`react-konva` or `react-leaflet`|For advanced polygon interaction|
|`tailwindcss`|Clean layout/styling|
|`Framer Motion`|Transitions, UX polish|
|`FastAPI` (optional now)|For backend REST API when ready|
