```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------
# Mock-up inputs (adjustable)
# ------------------------------
affected_pct = np.linspace(0, 50, 101)  # % of blast affected by voids/edges/wet & lost holes
budget_dig_rate = 1000                  # bcm/hr (budget)
observed_low = 700                      # bcm/hr (typical lower bound)
observed_high = 1100                    # bcm/hr (typical upper bound)

# ------------------------------
# Simple illustrative mapping
# ------------------------------
# Mean dig rate penalty: mild at low affected %, stronger as it grows
# (chosen to be plausible-looking for a mock-up, not calibrated)
penalty = 0.0035 * affected_pct + 0.00003 * affected_pct**2  # fractional penalty
mean = budget_dig_rate * (1 - penalty)

# Uncertainty widens with affected % (fan chart)
# Start at ±8% and widen to ~±25%
unc = 0.08 + 0.0034 * affected_pct
p90 = mean * (1 + unc)
p10 = mean * (1 - unc)

# Clamp to a realistic floor for this mock-up (optional)
floor = 450
mean = np.maximum(mean, floor)
p10 = np.maximum(p10, floor)

# ------------------------------
# Plot (single chart, more "impact")
# ------------------------------
plt.figure(figsize=(11, 6.5))

# Typical variability band (horizontal)
plt.fill_between(
    affected_pct,
    observed_low,
    observed_high,
    alpha=0.15,
    label="Typical operating range (700-1100 bcm/hr)"
)

# Confidence fan (p10-p90)
plt.fill_between(
    affected_pct,
    p10,
    p90,
    alpha=0.25,
    label="Predicted confidence band (P10-P90)"
)

# Mean (P50) line
plt.plot(affected_pct, mean, linewidth=2.0, label="Predicted mean (P50)")

# Budget line
plt.axhline(budget_dig_rate, linestyle="--", linewidth=1.5, label="Budget dig rate (1000 bcm/hr)")

# Labels
plt.xlabel("Proportion of blast affected by voids / edge effects / wet & lost holes (%)")
plt.ylabel("Dig rate (bcm/hr)")
plt.title("Dig Rate Penalty + Confidence vs Affected Blast Proportion")

# Add a few callouts (kept simple, high-signal)
for x in [0, 20, 40, 50]:
    idx = np.argmin(np.abs(affected_pct - x))
    plt.scatter([affected_pct[idx]], [mean[idx]], s=25)
    plt.text(
        affected_pct[idx] + 0.8,
        mean[idx],
        f"{int(round(mean[idx]))}",
        fontsize=9,
        va="center"
    )

plt.xlim(0, 50)
plt.ylim(400, 1200)
plt.legend(loc="lower left", frameon=True)
plt.tight_layout()
plt.show()
```

```python
# ------------------------------
# Plot 1: PSD shift mock-up
# ------------------------------
import numpy as np
import matplotlib.pyplot as plt

# Particle size range (mm)
x = np.linspace(1, 1500, 500)

# Simple illustrative S-curve (logistic-style CDF approximation)
def psd_curve(x, p50, sharpness):
    return 100 / (1 + (p50 / x)**sharpness)

# Baseline PSD
baseline = psd_curve(x, p50=250, sharpness=1.8)

# Risk-adjusted PSD (coarser + heavier tail)
risk_adjusted = psd_curve(x, p50=350, sharpness=1.4)

# Plot
plt.figure(figsize=(10, 6))
plt.plot(x, baseline, label="Baseline predicted PSD")
plt.plot(x, risk_adjusted, linestyle="--", label="Risk-adjusted PSD (30% affected)")

# Oversize threshold (example 800 mm)
oversize_threshold = 800
plt.axvline(oversize_threshold, linestyle=":", label="Oversize threshold (800 mm)")

plt.xlabel("Fragment size (mm)")
plt.ylabel("Cumulative passing (%)")
plt.title("Effect of Affected Zones on PSD (Coarse Tail Shift)")
plt.xscale("log")
plt.ylim(0, 100)
plt.legend()
plt.tight_layout()
plt.show()

# ------------------------------
# Plot 2: Dig rate fan chart
# ------------------------------

affected_pct = np.linspace(0, 50, 101)
budget = 1000
observed_low = 700
observed_high = 1100

# Mean penalty
penalty = 0.0035 * affected_pct + 0.00003 * affected_pct**2
mean = budget * (1 - penalty)

# Confidence widening
unc = 0.08 + 0.0034 * affected_pct
p90 = mean * (1 + unc)
p10 = mean * (1 - unc)

plt.figure(figsize=(11, 6))

# Observed operating range
plt.fill_between(affected_pct, observed_low, observed_high, alpha=0.15, label="Observed operating range (700-1100 bcm/hr)")

# Predicted confidence band
plt.fill_between(affected_pct, p10, p90, alpha=0.25, label="Predicted confidence band (P10-P90)")

# Mean line
plt.plot(affected_pct, mean, linewidth=2.0, label="Predicted mean (P50)")

# Budget line
plt.axhline(budget, linestyle="--", label="Budget dig rate (1000 bcm/hr)")

plt.xlabel("Proportion of blast affected by voids / edge effects / wet & lost holes (%)")
plt.ylabel("Dig rate (bcm/hr)")
plt.title("Dig Rate Penalty + Confidence vs Affected Blast Proportion")

plt.xlim(0, 50)
plt.ylim(400, 1200)
plt.legend(loc="lower left")
plt.tight_layout()
plt.show()
```

### PSD Shift Plot (Mechanism)

It clearly shows:

- Baseline fragmentation
- Coarser “risk-adjusted” fragmentation
- Oversize threshold

Engineers immediately understand:

- “Right tail gets heavier.”
- “Oversize proportion increases.”
- “This is how voids / edges / wet holes manifest physically.”

It explains cause, not just consequence.

### Dig Rate Fan Chart (Operational Impact)

This shows:

- Budget (1000 bcm/hr)
- Observed real-world range (700-1100 bcm/hr)
- Predicted mean drop
- Confidence widening

It communicates:

- Mean performance degrades.
- Variability increases.
- Risk of missing budget grows as affected proportion increases.

It connects directly to planning and scheduling decisions.

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------
# Plot 2: Column-based excavation zones + line overlay
# ------------------------------
categories = ["Good", "Reasonable", "Moderate effects", "Large effects"]
area_pct = np.array([55, 25, 15, 5])  # % of excavation area in each zone (example)

# Representative dig-rate impacts per zone (example, not calibrated)
# (Mean drops and uncertainty widens with severity)
mean = np.array([1030, 960, 850, 720])  # bcm/hr
p10  = np.array([950,  840,  680, 520])
p90  = np.array([1100, 1080, 1020, 920])

x = np.arange(len(categories))

fig, ax_area = plt.subplots(figsize=(11, 6.5))

# Bars: area proportion
ax_area.bar(x, area_pct)
ax_area.set_ylabel("Share of excavation area (%)")
ax_area.set_ylim(0, 70)
ax_area.set_xticks(x)
ax_area.set_xticklabels(categories, rotation=0)
ax_area.set_xlabel("Excavation condition class (based on voids / edges / wet & lost holes)")
ax_area.set_title("Area Mix by Condition Class with Dig Rate Mean + Confidence")

# Line + band: dig rate on secondary axis
ax_rate = ax_area.twinx()
ax_rate.plot(x, mean, linewidth=2.0, label="Predicted mean (P50)")
ax_rate.fill_between(x, p10, p90, alpha=0.25, label="Predicted confidence band (P10-P90)")

# Reference lines
budget = 1000
obs_low, obs_high = 700, 1100
ax_rate.axhline(budget, linestyle="--", linewidth=1.5, label="Budget (1000 bcm/hr)")
ax_rate.fill_between([-0.4, len(categories)-0.6], obs_low, obs_high, alpha=0.12, label="Observed range (700-1100 bcm/hr)")

ax_rate.set_ylabel("Dig rate (bcm/hr)")
ax_rate.set_ylim(400, 1200)

# Combined legend (from both axes)
handles1, labels1 = ax_area.get_legend_handles_labels()
handles2, labels2 = ax_rate.get_legend_handles_labels()
ax_rate.legend(handles1 + handles2, labels1 + labels2, loc="lower left")

plt.tight_layout()
plt.show()

# ------------------------------
# Plot 3: Weighted severity index (continuous) + dig rate fan
# ------------------------------
# Severity weights and example area mix (same as above)
weights = np.array([0.0, 0.5, 1.0, 2.0])
weighted_index = float(np.sum(weights * area_pct))  # "severity score" in %-units

# Build a continuous axis for demonstration (0 to 80 severity score)
sev = np.linspace(0, 80, 161)

# Map severity score -> mean dig rate + uncertainty (illustrative)
mean2 = 1000 * (1 - (0.0032 * sev + 0.00002 * sev**2))
unc2 = 0.08 + 0.0028 * (sev / 1.0)   # widens with severity
p90_2 = mean2 * (1 + unc2)
p10_2 = mean2 * (1 - unc2)

plt.figure(figsize=(11, 6.5))

# Observed range band
plt.fill_between(sev, obs_low, obs_high, alpha=0.15, label="Observed operating range (700-1100 bcm/hr)")

# Confidence band
plt.fill_between(sev, p10_2, p90_2, alpha=0.25, label="Predicted confidence band (P10-P90)")

# Mean line
plt.plot(sev, mean2, linewidth=2.0, label="Predicted mean (P50)")

# Budget line
plt.axhline(budget, linestyle="--", linewidth=1.5, label="Budget dig rate (1000 bcm/hr)")

# Mark the example blast's computed index
plt.axvline(weighted_index, linestyle=":", linewidth=1.8, label=f"Example blast index = {weighted_index:.1f}")
plt.text(weighted_index + 1.0, 440, "Example blast", fontsize=9, va="bottom")

plt.xlabel("Weighted oversize severity index (Σ weight × area%)")
plt.ylabel("Dig rate (bcm/hr)")
plt.title("Dig Rate Penalty + Confidence vs Weighted Severity Index")

plt.xlim(0, 80)
plt.ylim(400, 1200)
plt.legend(loc="upper left")
plt.tight_layout()
plt.show()
```

### Column-based excavation zones + line overlay

What it communicates (very clearly to D&B teams):

- Bars: “How much of the dig area is in each condition class”
- Line + band: “What dig rate we expect for each class, and how uncertain that is”
- Plus you still show budget (1000 bcm/hr) and the observed operating band (700-1100 bcm/hr)

Why it works:

- It matches how engineers think: “Most of this blast is fine, but we’ve got a bad corner / perimeter rows / old workings strip.”
- It’s a nice bridge between spatial reality and performance prediction.

When it can get noisy:
- If you try to show too many classes (keep it to 4-5 max).
- If you add too many overlays (don’t add PSD here).

### Weighted severity index (Σ weight × area%) + fan chart

What it communicates:

- A single scalar that represents the blast’s overall “oversize severity exposure”
- You get back to a clean “one axis → one outcome” relationship:
	- higher index → lower mean dig rate + wider confidence band
- You can mark the current blast’s index with a vertical line.

Why it works:

- It’s clean and scalable (you can add more drivers later without changing the chart type).
- It’s excellent for comparing blasts: Blast A index 20 vs Blast B index 45.

Where it’s weaker:

- You lose some intuitive spatial sense unless you pair it with the bar chart or a small table showing the category breakdown.
