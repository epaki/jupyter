```python
# Simple chatbot program

precepts = ["help", "hi", "how are you", "hello", "tea", "cafe", "no", "thank you", "thanks", "goodbye", "bye"]
# Define a list of known questions and their responses

responses = [
    "Sure thing, how can i help you today?",
    "Hi there! How can I assist you today?",
    "I'm well thanks - how about you? How can I assist you today?",
    "Hello! How can I assist you today?",
    "Sure, where would you like to it? A cafe in the campus maybe?",
    "There are 2 cafes - in main building and near the lecture halls.",
    "Is there anything else I can help you with",
    "You are welcome.",
    "You're welcome.",
    "Goodbye! Have a great day!"
    "Bye! Have a top day!"
]
```

```python
# Function to get user input and respond
def chatbot():

    # User welcome and termination instruction.

    # Main loop:
    while True:
        print("Type 'exit' anytime to stop.\n")
        # 1. Get user input
        user_input = input("You: ").lower()
        
        # 2. Check if the user wants to exit
        if user_input == "exit" or user_input == "bye":
            print("Chatbot: Goodbye! 👋")
            break
        
        # 3. Search for a pattern in the precepts. And fetch a predefined responses
        found = False
        
        #   3.1 Check if any of precepts are there in the input
        for i, pattern in enumerate(precepts):
            if pattern in user_input:
                #   3.2 Return the index
                #   3.3 Use index to find repsonse
                print("Chatbot: ", responses[i])
                
                #   3.4 Break the loop
                found = True
                break
        
        if not found:
            print("Chatbot: Sorry, I don't understand that. Try asking something else.")
        
# Run the chatbot
chatbot()
```

### 2.1 Lists -> Dictionaries

To convert lists to dictionaries, let's use the lists from previous activity.

```python
# Write a function that takes in precept categories and responses lists and returns a mapping between them.

# - Use ZIP function to combine two independent lists (precepts and responses) and create a new map between them.
# - Search on the ZIP function and how to use it in a for loop
# - Run the main loop and check that this function works

# ___ Write your code here _____

import re
import random


def zip_responses(precepts, responses):
    """
    Takes a list of precept categories and a list of response lists.
    Returns a dictionary mapping each precept to its possible responses.
    """
    mapping = {}

    for precept, response in zip(precepts, responses):
        mapping[precept] = response

    return mapping
```

- Match complete words using regular expressions.

```python
# Create 5 RE patterns

# Dictionary name: "re_patterns"

# ___ Write your code here _____
re_patterns = {
    r"^(help|support|assist|assistance|menu)$": "help",
    r"\b(hi|hello|hey|howdy|yo)\b": "greetings",
    r"\b(bye|goodbye|farewell|cya|later)\b": "goodbye",
    r"\b(thanks|thank you|cheers|appreciate|ta)\b": "thanks",
    r"\b(?:i would like|i want|can i have|may i have|please give me)\s+(?:some\s+)?(?P<drink>tea|coffee|water|juice|soda)\b": "tea_request",
    r"\b(?:my name is|i am called|i'm called|call me)\s+(?P<name>[A-Za-z]+)\b": "name_response",
    r"\b(?:my favou?rite colou?r is|colou?r is|i like)\s+(?P<colour>[A-Za-z]+)\b": "colour_response",
    r"\b(?:i am|i'm)\s+(?P<age>\d{1,3})\s*(?:years?\s*old)?\b": "age_response",
    r"\b(?P<email>[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,})\b": "email_response",
}

re_patterns
```

### 2.3 Function : Detect pattern from input

---

Use regular expressions + dictionary from before to detect a pattern from user input. 

**Function input** : User Input ("Hi, how are you?")  <br>
**Output**         : Response_type ("Greetings") 

---

```python
## Function name: "detect_pattern"

## Create the detect_pattern function with one argument : user_input
## 1. Loop through each key in the dictionary:
## 2. Write if-else conditions to check for pattern match
##    2.1 For the if condition, use the re.search to check if the pattern exists in user input 
##    2.2 If there is a match, return the response type (e.g., "greetings" or "bye" or "happy")
##    2.3 If none of the conditions match, return "default" which will be the default response of the chatbot
##    Note: use else for last conditon

## TIP: To avoid multiple conditions from matching, use a break statement. 

# Give the function a characteristic name (e.g., detect_pattern) 
# ___ Write your code here _____

user_state = {
    "name": "",
    "colour": "",
    "age": "",
    "email": "",
    "drink": ""
}


def detect_pattern(user_input):
    """
    Detects a pattern in the user's input using regular expressions.
    Returns the response category, such as 'greetings' or 'default'.

    If the regex contains named groups, the matched values are stored in user_state.
    """
    cleaned_input = user_input.strip()

    for pattern, response_type in re_patterns.items():
        result = re.search(pattern, cleaned_input, re.IGNORECASE)

        if result:
            # Store values from any named capture groups, e.g. name, colour, age, email or drink.
            for key, value in result.groupdict().items():
                if value:
                    if key == "name":
                        user_state[key] = value.title()
                    elif key in ["colour", "drink"]:
                        user_state[key] = value.lower()
                    else:
                        user_state[key] = value

            return response_type

    return "default"
```

### 2.4 Function : Retrieve response

---

Use the output of pattern detection function (above) to trigger a chatbot response. 

In the function, use the "responses" dictionary from above. 


Follow the steps to write function. The function returns a response (e.g., "hello", or "how are you").

**Function input** : User Input ("Hi, how are you?")  <br>
**Detect pattern** : Response_type ("Greetings")  <br>
**Output**         : One of ["hello", "how are you?", etc...]

---

```python
## Function name: "chatbot_response" 

# Write a function to return a responses by indexing dictionary with response categories.

## Create the chatbot_response function with one argument : user_input
## 1. Call the detect_pattern function and store the response in variable (e.g., detected_type)
## 2. Write if-else conditions to check for pattern match
##    2.1 For the if condition, check if the detected_type is one of the keys of the dictionary
##    2.2 If there is a match, return the response from zipped responses
##    2.3 If none of the conditions, use "default" response

# Give the function a characteristic name (e.g., detect_pattern) 
# ___ Write your code here _____

def chatbot_response(user_input):
    """
    Uses detect_pattern to identify the response type.
    Then retrieves and formats an appropriate response from the responses dictionary.
    """
    detected_type = detect_pattern(user_input)

    if detected_type not in responses:
        detected_type = "default"

    response = random.choice(responses[detected_type])

    # These values are used only when the chosen template contains placeholders.
    format_values = {
        "name": user_state.get("name", ""),
        "colour": user_state.get("colour", ""),
        "age": user_state.get("age", ""),
        "email": user_state.get("email", ""),
        "drink": user_state.get("drink", "")
    }

    return response.format(**format_values)
```

```python
## A reminder for what the main loop looked like.
def chat():
    """Interactive chatbot loop. Run chat() when you are ready to record your video."""
    print("Chatbot: Hello! Type 'exit' or 'bye' to stop.\n")

    while True:
        user_input = input("You: ").strip()

        if re.fullmatch(r"exit|bye|goodbye", user_input, re.IGNORECASE):
            print("Chatbot: Goodbye! 👋")
            break

        bot_response = chatbot_response(user_input)
        print(f"Chatbot: {bot_response}\n")


def scripted_demo():
    """Prints a short demonstration without requiring keyboard input."""
    demo_inputs = [
        "help",                                
        "Hello there",                         
        "My name is Bob",                     
        "My favourite colour is blue",        
        "I am 21 years old",                   
        "Please contact me at bob888@uol.com",  
        "I want some tea",                     
        "This sentence should not match anything"   
    ]

    random.seed(7)  # Keeps the demo output stable for screenshots/videos.

    for message in demo_inputs:
        print(f"You: {message}")
        print(f"Chatbot: {chatbot_response(message)}\n")

# Start
# chat()
scripted_demo()
```

### Strings -> regex

```python
# To convert patterns from strings to REs, we have to look 
# at each pattern and concatenate them with a pipe symbol.

# Loop thorugh patterns list and tags.
 # For each pattern in list, add an "r" symbol
 # Create a dictionary with pattern and tags

patterns = ["hello","hi","hiya"]

r"hello|hi|hiya"

r"|".join(patterns)
```

### Dictionary 1: Patterns -> intent 

```python
# Map one combined regular expression to each intent tag.
def store_patterns(tags, patterns):
    regex2int = {}

    for tag, pattern_list in zip(tags, patterns):
        regex_pattern = r"(?<!\w)(?:" + "|".join(pattern_list) + r")(?!\w)"
        regex2int[regex_pattern] = tag

    return regex2int


regex2int = store_patterns(tags, patterns)
regex2int
```

### Dictionary 2: intent -> responses


TIP: For any feature, write this as a function. (e.g. store_responses)

```python
# Map every intent tag to its possible chatbot responses.
def store_responses(tags, responses):
    int2res = {}

    for tag, response_list in zip(tags, responses):
        int2res[tag] = response_list

    return int2res


int2res = store_responses(tags, responses)
int2res
```

### 3.3 Integrating file-based data into chatbot logic

Follow the code templates provided below to understand how files can be integrated with the Chatbot logic. 

*Fill in appropriate code segments.*


```python
# Build both dictionaries from the data read from intents.json.
def load_files(filename="intents.json"):
    intents_json = read_json(filename)
    tags = []
    patterns = []
    responses = []

    for intent in intents_json:
        tags.append(intent['tag'])
        patterns.append(intent['patterns'])
        responses.append(intent['responses'])

    regex2int = store_patterns(tags, patterns)
    int2res = store_responses(tags, responses)
    return regex2int, int2res


regex2int, int2res = load_files()
print("File-based chatbot knowledge loaded.")
```

```python
# Load the file-based dictionaries, then search the user's input for a match.
def generate_response(user_input):
    regex2int, int2res = load_files()

    for regex_pattern, intent in regex2int.items():
        match = re.search(regex_pattern, user_input, re.IGNORECASE)

        if match:
            response = random.choice(int2res[intent])
            response = response.format(**match.groupdict())
            print(f"Chatbot: {response}")
            return True

    return False


# A short, repeatable check before using the interactive chatbot.
random.seed(8)
for message in ["Hello", "My name is Eden", "See you later", "What is the weather?"]:
    print(f"You: {message}")
    response_found = generate_response(message)
    if response_found is False:
        print("Chatbot: I'm sorry, could you rephrase that again?")
    print()
```
