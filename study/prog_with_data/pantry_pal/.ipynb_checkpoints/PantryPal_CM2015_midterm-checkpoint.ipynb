# PantryPal - recipe helper chatbot

This notebook contains my midterm chatbot project for CM2015 Programming with Data.

**Domain:** PantryPal is a small recipe helper. It asks about ingredients, meal type, time limits and dietary preference, then suggests a simple recipe from a local JSON file.

I chose this topic because it is practical but still small enough to test properly. The chatbot is not meant to be a full recipe website or nutrition tool; it just demonstrates regex matching, text preprocessing, dictionaries, lists, JSON file use and a simple main loop.

## 1. Imports and file location

The project uses a separate `intents.json` file for the chatbot data. The path is relative, so the notebook runs as long as the JSON file is in the same folder as this notebook.

```python
from pathlib import Path
import json
import random
import re

from nltk.stem import SnowballStemmer

INTENTS_FILE = Path("intents.json")
random.seed(13)
```

## 2. Loading the chatbot data

The JSON file stores the regex patterns, response templates and a small recipe list. I build two dictionaries from it:

- `pattern2intent`, which maps each regex pattern to an intent tag
- `intent2responses`, which maps each intent tag to possible responses

```python
def load_intents(filename=INTENTS_FILE):
    """Load the external JSON file used by the chatbot."""
    filename = Path(filename)
    with open(filename, "r", encoding="utf-8") as f:
        return json.load(f)


def build_intent_maps(data):
    """Create lookup dictionaries from the JSON structure."""
    pattern2intent = {}
    intent2responses = {}

    for item in data["intents"]:
        tag = item["tag"]
        intent2responses[tag] = item["responses"]
        for pattern in item["patterns"]:
            pattern2intent[pattern] = tag

    return pattern2intent, intent2responses


data = load_intents()
pattern2intent, intent2responses = build_intent_maps(data)

print(data["bot_name"])
print(f"Loaded {len(data['intents'])} intents, {len(pattern2intent)} regex patterns and {len(data['recipes'])} recipes.")
```

## 3. Pre-processing text

This section cleans user text before I use it for token checks and recipe matching. I also keep the original text for regex matching because punctuation can matter in phrases such as `I've got`.

The preprocessing is deliberately simple: lowercase text, remove punctuation that is not useful for matching, split into tokens, remove common stop words and make a stemmed version of the main tokens. I use stemming carefully because very aggressive stemming can make results harder to understand.

```python
stemmer = SnowballStemmer("english")

STOP_WORDS = {
    "a", "an", "and", "are", "can", "for", "have", "i", "in", "is", "it",
    "me", "my", "of", "on", "or", "please", "the", "to", "what", "with", "you"
}


def clean_text(text):
    """Lowercase text and remove punctuation/special characters used as noise."""
    text = text.lower().replace("’", "'")
    text = re.sub(r"[^a-z0-9\s'-]+", " ", text)
    text = re.sub(r"[-']", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text


def tokenise(text):
    """Split cleaned text into words and numbers."""
    return re.findall(r"[a-z]+|\d+", clean_text(text))


def preprocess(text):
    """Return the cleaned text, tokens and stemmed tokens for inspection."""
    tokens = tokenise(text)
    important_tokens = [word for word in tokens if word not in STOP_WORDS]
    stems = [stemmer.stem(word) for word in important_tokens]
    return {
        "raw": text,
        "clean": clean_text(text),
        "tokens": tokens,
        "important_tokens": important_tokens,
        "stems": stems,
    }


example = "Hello!!! I've got pasta, tomato & garlic."
preprocess(example)
```

## 4. Regex intent matching

Each pattern is stored in the JSON file. The matching function searches with `re.IGNORECASE` so that the chatbot treats `Hello`, `hello` and `HELLO` as the same intent. Named regex groups, such as `(?P<name>...)`, are used to capture details that can be saved into memory.

```python
def find_intent(user_text, pattern2intent):
    """Return the first intent whose regex pattern matches the user text."""
    cleaned = clean_text(user_text)

    for pattern, intent in pattern2intent.items():
        match = re.search(pattern, user_text, flags=re.IGNORECASE)
        if match:
            return intent, match

        # A second pass on cleaned text helps with extra punctuation.
        match = re.search(pattern, cleaned, flags=re.IGNORECASE)
        if match:
            return intent, match

    return "fallback", None


for text in ["HELLO!", "My name is Bob", "I need dinner in 20 mins"]:
    intent, match = find_intent(text, pattern2intent)
    print(f"{text!r} -> {intent}")
```

## 5. Memory and recipe matching

The chatbot keeps a small memory dictionary during one conversation. This is used for details such as name, diet, meal type, time limit and ingredients. The recipe matching is not complicated: recipes get a score for matching ingredients, diet, meal type and time limit.

```python
def fresh_memory():
    return {
        "name": "there",
        "diet": None,
        "servings": None,
        "meal": None,
        "max_minutes": None,
        "ingredients": [],
        "item": "that ingredient",
    }


def normalise_diet(value):
    if not value:
        return None
    value = clean_text(value).replace(" ", "-")
    if value == "no-meat":
        return "vegetarian"
    if value == "gluten-free":
        return "gluten-free"
    return value


def add_unique_items(existing, new_items):
    for item in new_items:
        if item not in existing:
            existing.append(item)
    return existing


def extract_ingredients(text, data):
    """Find known ingredients in the user's message using the aliases in the JSON file."""
    cleaned = clean_text(text)
    found = []

    # Longer aliases are checked first so phrases such as 'soy sauce' win over 'soy'
    aliases = sorted(data.get("ingredient_aliases", {}).items(), key=lambda pair: len(pair[0]), reverse=True)
    for alias, canonical in aliases:
        alias_clean = clean_text(alias)
        pattern = r"\b" + re.escape(alias_clean) + r"\b"
        if re.search(pattern, cleaned):
            found.append(canonical)

    return add_unique_items([], found)


def diet_is_ok(recipe, diet):
    if diet is None:
        return True
    recipe_diets = set(recipe["diet"])
    if diet == "vegetarian":
        return "vegetarian" in recipe_diets or "vegan" in recipe_diets
    return diet in recipe_diets


def update_memory(user_text, intent, match, memory, data):
    """Update the memory dictionary using regex groups and extra keyword checks."""
    groups = match.groupdict() if match else {}

    if groups.get("name"):
        memory["name"] = groups["name"].strip().title()
    if groups.get("diet"):
        memory["diet"] = normalise_diet(groups["diet"])
    if groups.get("servings"):
        memory["servings"] = int(groups["servings"])
    if groups.get("minutes"):
        memory["max_minutes"] = int(groups["minutes"])
    if groups.get("meal"):
        memory["meal"] = groups["meal"].lower()
    if groups.get("item"):
        memory["item"] = groups["item"].lower()

    # These extra checks make the chatbot less dependent on one exact phrase
    diet_match = re.search(r"\b(vegetarian|vegan|gluten[- ]?free|no meat)\b", user_text, flags=re.IGNORECASE)
    if diet_match:
        memory["diet"] = normalise_diet(diet_match.group(1))

    meal_match = re.search(r"\b(breakfast|lunch|dinner|snack)\b", user_text, flags=re.IGNORECASE)
    if meal_match:
        memory["meal"] = meal_match.group(1).lower()

    time_match = re.search(r"\b(?:under|within|in|less than)?\s*(\d{1,3})\s*(?:mins?|minutes?)\b", user_text, flags=re.IGNORECASE)
    if time_match:
        memory["max_minutes"] = int(time_match.group(1))

    serving_match = re.search(r"\b(?:for|serves?|feeding)\s+(\d{1,2})\b|\b(\d{1,2})\s+(?:people|person|servings|portions)\b", user_text, flags=re.IGNORECASE)
    if serving_match:
        memory["servings"] = int(next(group for group in serving_match.groups() if group))

    ingredients = extract_ingredients(user_text, data)
    if ingredients:
        memory["ingredients"] = add_unique_items(memory["ingredients"], ingredients)

    return memory


def score_recipe(recipe, memory):
    """Score one recipe against the current memory."""
    if not diet_is_ok(recipe, memory.get("diet")):
        return -100

    score = 0
    recipe_ingredients = set(recipe["ingredients"])
    user_ingredients = set(memory.get("ingredients", []))

    matches = recipe_ingredients.intersection(user_ingredients)
    score += 3 * len(matches)

    if user_ingredients and not matches:
        score -= 1

    if memory.get("diet"):
        score += 2

    if memory.get("meal"):
        if memory["meal"] in recipe["meal"]:
            score += 2
        else:
            score -= 1

    if memory.get("max_minutes"):
        if recipe["minutes"] <= memory["max_minutes"]:
            score += 2
        else:
            score -= 3

    return score


def recommend_recipes(data, memory, limit=2):
    scored = []
    for recipe in data["recipes"]:
        score = score_recipe(recipe, memory)
        if score > -100:
            scored.append((score, recipe["minutes"], recipe))

    scored.sort(key=lambda item: (-item[0], item[1], item[2]["name"]))
    return [recipe for score, minutes, recipe in scored[:limit]]
```

## 6. Response generation

Responses are selected from the JSON file. Some responses use placeholders such as `{name}` and `{diet}`. For recipe requests, the chatbot adds a short recipe recommendation based on the current memory.

```python
def memory_context(memory, data):
    recipe_names = ", ".join(recipe["name"] for recipe in data["recipes"])
    ingredients = ", ".join(memory.get("ingredients", [])) or "none yet"
    return {
        "name": memory.get("name") or "there",
        "diet": memory.get("diet") or "not set",
        "servings": memory.get("servings") or "not set",
        "meal": memory.get("meal") or "not set",
        "max_minutes": memory.get("max_minutes") or "not set",
        "ingredients": ingredients,
        "item": memory.get("item") or "that ingredient",
        "recipe_list": recipe_names,
    }


def recipe_response(data, memory):
    recommendations = recommend_recipes(data, memory)
    if not recommendations:
        return "I could not find a recipe that fits those details. Try removing one restriction or listing different ingredients."

    top = recommendations[0]
    matched = set(top["ingredients"]).intersection(memory.get("ingredients", []))
    matched_text = f" It matches: {', '.join(sorted(matched))}." if matched else ""
    serving_text = f" for {memory['servings']} serving(s)" if memory.get("servings") else ""

    answer = (
        f"I would suggest {top['name']}{serving_text}. It takes about {top['minutes']} minutes."
        f"{matched_text} {top['method']}"
    )

    if len(recommendations) > 1:
        answer += f" Another possible option is {recommendations[1]['name']}."

    return answer


def generate_response(intent, intent2responses, data, memory):
    if intent == "reset":
        memory.clear()
        memory.update(fresh_memory())

    context = memory_context(memory, data)
    template = random.choice(intent2responses.get(intent, intent2responses["fallback"]))
    base_response = template.format(**context)

    if intent in {"ingredient_recipe", "recipe_suggestion", "meal_type"}:
        return recipe_response(data, memory)

    return base_response


def reply(user_text, data, memory=None):
    """Process one user message and return the chatbot response."""
    if memory is None:
        memory = fresh_memory()

    pattern2intent, intent2responses = build_intent_maps(data)
    intent, match = find_intent(user_text, pattern2intent)
    update_memory(user_text, intent, match, memory, data)
    return generate_response(intent, intent2responses, data, memory)


def should_quit(user_text):
    return clean_text(user_text) in {"exit", "quit"}
```

## 7. Main loop

The main loop keeps asking for input until the user types `exit` or `quit`. I've left the actual call to `chat()` commented out so that the notebook can run from top to bottom without getting stuck waiting for input.

```python
def chat(filename=INTENTS_FILE):
    data = load_intents(filename)
    memory = fresh_memory()

    print("PantryPal: Hello. Tell me some ingredients, or ask for breakfast, lunch or dinner ideas.")
    while True:
        user_text = input("You: ")
        if should_quit(user_text):
            print("PantryPal: Goodbye.")
            break
        print("PantryPal:", reply(user_text, data, memory))


# Remove the comment from the chat() call to run this manually
chat()
```

## 8. Short demo transcript

This gives a quick demonstration without needing interactive input.

```python
demo_memory = fresh_memory()
demo_messages = [
    "Hello",
    "My name is Bob",
    "I am vegetarian",
    "I have pasta, tomato and garlic and need dinner in 20 mins",
    "What do you remember?",
]

for message in demo_messages:
    print("User:", message)
    print("PantryPal:", reply(message, data, demo_memory))
    print()
```

## 9. Tests

The tests focus on the main behaviours I wanted to check: preprocessing, regex intent matching, memory updates, recipe matching and the exit condition

```python
def run_tests():
    test_data = load_intents()
    test_patterns, _ = build_intent_maps(test_data)

    processed = preprocess("Hello!!! I've got pasta, tomato & garlic.")
    assert processed["clean"] == "hello i ve got pasta tomato garlic"
    assert "pasta" in processed["tokens"]
    assert "garlic" in processed["important_tokens"]

    intent, match = find_intent("My name is Maya", test_patterns)
    assert intent == "set_name"
    memory = fresh_memory()
    update_memory("My name is Maya", intent, match, memory, test_data)
    assert memory["name"] == "Maya"

    memory = fresh_memory()
    answer = reply("I have pasta, tomato and garlic and need dinner in 20 mins", test_data, memory)
    assert "Tomato pasta" in answer
    assert memory["meal"] == "dinner"
    assert memory["max_minutes"] == 20
    assert set(["pasta", "tomato", "garlic"]).issubset(memory["ingredients"])

    memory = fresh_memory()
    reply("I am vegan", test_data, memory)
    answer = reply("Suggest lunch with chickpeas and tomato", test_data, memory)
    assert memory["diet"] == "vegan"
    assert "Chickpea salad" in answer
    assert "Tuna wrap" not in answer

    assert should_quit("quit")
    assert should_quit(" exit ")
    assert not should_quit("goodbye")

    return "All tests passed."


run_tests()
```

## 10. Notes on limitations

PantryPal only has a small recipe list, so it cannot cover every possible ingredient. It also uses regex matching rather than a large language model, so unusual phrasing may fall back to the default response. The advantage is that the behaviour is easy to inspect and test, which suits the scope of this coursework.
