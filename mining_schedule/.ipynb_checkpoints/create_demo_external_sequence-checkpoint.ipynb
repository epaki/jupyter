# Create demo external scheduler sequence

This notebook generates a synthetic external scheduler sequence for the Blast Master Scenario Simulation demo. It uses the supplied blast-master polygons, copper block model, and drill/charge design files.

The output is intended for demonstration and prototype testing. It is not a real commercial scheduler output.

## Outputs

- `demo_external_scheduler_sequence.csv` — flat pattern-level sequence for import.
- `demo_external_scheduler_periods.csv` — day-level period summaries.
- `demo_external_scheduler_schedule_asset.json` — schedule asset matching the current prototype schema as closely as possible.
- `README_demo_external_sequence.md` — usage notes.

```
import json
import math
import zipfile
from datetime import datetime, timedelta, timezone
from pathlib import Path
from statistics import mode

import numpy as np
import pandas as pd
from shapely.geometry import Point, Polygon

# Run this notebook from the folder containing the source demo files, or leave them in /mnt/data.
CWD = Path.cwd()
BASE = CWD if (CWD / 'blast_master.csv').exists() else Path('/mnt/data')
BLAST_MASTER = BASE / 'blast_master.csv'
BLOCK_MODEL = BASE / 'block_model_copper.csv'
DRILL_CHARGE_ZIP = BASE / 'drill_charge_designs.zip'
OUT_DIR = BASE / 'demo_external_sequence_package'
OUT_DIR.mkdir(exist_ok=True)

START = datetime(2026, 6, 1, 6, 0, 0, tzinfo=timezone.utc)
BENCH_HEIGHT_M = 15.0
BLOCK_DX_M = 10.0
BLOCK_DY_M = 10.0
BLOCK_VOL_M3 = BLOCK_DX_M * BLOCK_DY_M * BENCH_HEIGHT_M
CU_CUTOFF_PCT = 0.25
BASE_LOAD_RATE_TPH = 1400.0
AVG_TRUCK_PAYLOAD_T = 240.0
BASE_HAUL_DISTANCE_M = 1350.0


def clamp(v, lo, hi):
    return max(lo, min(hi, v))


def safe_float(v, default=None):
    try:
        if pd.isna(v):
            return default
        return float(v)
    except Exception:
        return default


def polygon_area(points):
    poly = Polygon(points)
    if not poly.is_valid:
        poly = poly.buffer(0)
    return poly


def weighted_mean(values, weights):
    values = np.asarray(values, dtype=float)
    weights = np.asarray(weights, dtype=float)
    mask = np.isfinite(values) & np.isfinite(weights) & (weights > 0)
    if not mask.any():
        return None
    return float(np.average(values[mask], weights=weights[mask]))


def top_value(series, default='unknown'):
    s = series.dropna().astype(str)
    if s.empty:
        return default
    return str(s.value_counts().idxmax())


blast = pd.read_csv(BLAST_MASTER)
blocks = pd.read_csv(BLOCK_MODEL)

# Load drill and charge designs from the ZIP.
with zipfile.ZipFile(DRILL_CHARGE_ZIP) as z:
    drill_frames = []
    charge_frames = []
    for name in z.namelist():
        if name.startswith('drill_designs/') and name.endswith('.csv'):
            with z.open(name) as f:
                drill_frames.append(pd.read_csv(f))
        if name.startswith('charge_designs/') and name.endswith('.csv'):
            with z.open(name) as f:
                charge_frames.append(pd.read_csv(f))

drill = pd.concat(drill_frames, ignore_index=True)
charge = pd.concat(charge_frames, ignore_index=True)

# Aggregate drill and charge design data.
drill_agg = drill.groupby('pattern_id').agg(
    hole_count=('hole_id', 'count'),
    drill_metres_m=('length_m', 'sum'),
    burden_m=('burden_m', 'mean'),
    spacing_m=('spacing_m', 'mean'),
    hole_diameter_mm=('hole_diameter_mm', 'mean'),
    design_explosive_kg=('explosive_mass_kg', 'sum'),
    powder_factor_kg_m3=('powder_factor_kg_m3', 'mean'),
    energy_intensity_mj_m3=('energy_intensity_mj_m3', 'mean'),
    predicted_p50_mm=('predicted_p50_mm', 'mean'),
    predicted_p80_mm=('predicted_p80_mm', 'mean'),
    predicted_p95_mm=('predicted_p95_mm', 'mean'),
    target_p80_mm=('target_p80_mm', 'mean'),
    hardness_index=('hardness_index', 'mean'),
    grade_index=('grade_index', 'mean'),
    energy_requirement_index=('energy_requirement_index', 'mean'),
    drill_density_tpm3=('density_tpm3', 'mean'),
    drill_lithology=('lithology', lambda s: top_value(s)),
    drill_domain=('mineral_domain', lambda s: top_value(s)),
    drill_digability=('digability', lambda s: top_value(s)),
).reset_index()

charge_agg = charge.groupby('pattern_number').agg(
    charge_holes=('hole_id', 'count'),
    explosive_kg=('design_explosive_kg', 'sum'),
    avg_stemming_m=('design_stemming_m', 'mean'),
    avg_explosive_column_m=('design_explosive_column_m', 'mean'),
    avg_hole_length_m=('design_hole_length', 'mean'),
).reset_index().rename(columns={'pattern_number': 'pattern_id'})

design_agg = drill_agg.merge(charge_agg, on='pattern_id', how='left')
design_agg['explosive_kg'] = design_agg['explosive_kg'].fillna(design_agg['design_explosive_kg'])

# Build pattern geometry and block-model summaries.
block_points = list(zip(blocks['x'].to_numpy(), blocks['y'].to_numpy()))
blocks['block_tonnes'] = BLOCK_VOL_M3 * blocks['density_tpm3']
blocks['is_ore'] = (blocks['cu_pct'] >= CU_CUTOFF_PCT) & (~blocks['mineral_domain'].isin(['waste']))

pattern_rows = []
for pid, group in blast.groupby('pattern_id'):
    g = group.sort_values('point_id')
    poly = polygon_area(g[['x', 'y']].to_numpy())
    centroid = poly.centroid
    inside = np.array([poly.contains(Point(x, y)) or poly.touches(Point(x, y)) for x, y in block_points])
    sub = blocks.loc[inside].copy()

    if len(sub) == 0:
        avg_density = 2.7
        estimated_tonnes = poly.area * BENCH_HEIGHT_M * avg_density
        ore_fraction = 0.0
        ore_tonnes = 0.0
        waste_tonnes = estimated_tonnes
        avg_cu = None
        rec_cu = None
        contained_cu_t = None
        recovered_cu_t = None
        avg_bwi = avg_ucs = avg_axb = avg_dwi = None
        digability = 'unknown'
        lithology = 'unknown'
        primary_domain = 'unknown'
        structure_domain = 'unknown'
        wet_fraction = None
    else:
        total_block_tonnes = sub['block_tonnes'].sum()
        avg_density = weighted_mean(sub['density_tpm3'], sub['block_tonnes']) or 2.7
        estimated_tonnes = poly.area * BENCH_HEIGHT_M * avg_density
        ore_mask = sub['is_ore']
        ore_fraction = float(sub.loc[ore_mask, 'block_tonnes'].sum() / total_block_tonnes) if total_block_tonnes > 0 else 0.0
        ore_tonnes = estimated_tonnes * ore_fraction
        waste_tonnes = max(0.0, estimated_tonnes - ore_tonnes)
        if ore_mask.any():
            avg_cu = weighted_mean(sub.loc[ore_mask, 'cu_pct'], sub.loc[ore_mask, 'block_tonnes'])
            rec_cu = weighted_mean(sub.loc[ore_mask, 'rec_cu'], sub.loc[ore_mask, 'block_tonnes'])
        else:
            avg_cu = weighted_mean(sub['cu_pct'], sub['block_tonnes'])
            rec_cu = weighted_mean(sub['rec_cu'], sub['block_tonnes'])
        contained_cu_t = ore_tonnes * (avg_cu or 0.0) / 100.0
        recovered_cu_t = contained_cu_t * (rec_cu or 0.0)
        avg_bwi = weighted_mean(sub['BWi_kWht'], sub['block_tonnes'])
        avg_ucs = weighted_mean(sub['UCS_MPa'], sub['block_tonnes'])
        avg_axb = weighted_mean(sub['Axb'], sub['block_tonnes'])
        avg_dwi = weighted_mean(sub['DWi_kWhm3'], sub['block_tonnes'])
        digability = top_value(sub['digability'])
        lithology = top_value(sub['lithology'])
        primary_domain = top_value(sub['mineral_domain'])
        structure_domain = top_value(sub['structure_domain'])
        wet_fraction = float(sub['wet'].mean()) if 'wet' in sub else None

    pattern_rows.append({
        'pattern_id': int(pid),
        'pattern_key': f'A-185-08-{int(pid):02d}',
        'pattern_type': str(group['pattern_type'].iloc[0]),
        'bench': 'RL185-200',
        'phase': 'Demo Phase A',
        'pushback': 'Pit A PB-01',
        'area_m2': float(poly.area),
        'perimeter_m': float(poly.length),
        'centroid_x': float(centroid.x),
        'centroid_y': float(centroid.y),
        'block_count': int(len(sub)),
        'estimated_tonnes': float(estimated_tonnes),
        'ore_tonnes': float(ore_tonnes),
        'waste_tonnes': float(waste_tonnes),
        'average_grade': None if avg_cu is None else float(avg_cu),
        'rec_cu': None if rec_cu is None else float(rec_cu),
        'contained_cu_t': None if contained_cu_t is None else float(contained_cu_t),
        'recovered_cu_t': None if recovered_cu_t is None else float(recovered_cu_t),
        'density_tpm3': float(avg_density),
        'BWi_kWht': None if avg_bwi is None else float(avg_bwi),
        'UCS_MPa': None if avg_ucs is None else float(avg_ucs),
        'Axb': None if avg_axb is None else float(avg_axb),
        'DWi_kWhm3': None if avg_dwi is None else float(avg_dwi),
        'digability': digability,
        'primary_mineral_domain': primary_domain,
        'primary_lithology': lithology,
        'structure_domain': structure_domain,
        'wet_fraction': wet_fraction,
        'ore_fraction': float(ore_fraction),
    })

patterns = pd.DataFrame(pattern_rows).merge(design_agg, on='pattern_id', how='left')

# Derived planning metrics.
patterns['hardness_factor'] = patterns['BWi_kWht'].fillna(patterns['BWi_kWht'].median()) / patterns['BWi_kWht'].median()
patterns['hardness_factor'] = patterns['hardness_factor'].clip(0.75, 1.45)
patterns['powder_factor_kg_t'] = patterns['explosive_kg'] / patterns['estimated_tonnes'].replace(0, np.nan)
patterns['load_haul_duration_hr'] = patterns['estimated_tonnes'] / BASE_LOAD_RATE_TPH
patterns['truck_hours'] = (patterns['estimated_tonnes'] / AVG_TRUCK_PAYLOAD_T) * (2 * (BASE_HAUL_DISTANCE_M / 1000.0) / 18.0)
patterns['loader_hours'] = patterns['load_haul_duration_hr']
patterns['drill_duration_hr'] = patterns['drill_metres_m'] / 70.0
patterns['charge_duration_hr'] = patterns['explosive_kg'] / 4800.0
patterns['total_duration_hr'] = patterns['drill_duration_hr'] + patterns['charge_duration_hr'] + patterns['load_haul_duration_hr']
patterns['haul_distance_m'] = BASE_HAUL_DISTANCE_M + (patterns['centroid_y'] - patterns['centroid_y'].min()) * 0.35 + (patterns['centroid_x'] - patterns['centroid_x'].mean()).abs() * 0.15

# Destination routing for demo. Keep enough variety to show scenario value.
def destination(row):
    if row['pattern_type'] in ['Ramp']:
        return 'Development / access material'
    if row['pattern_type'] in ['Slot']:
        return 'ROM pad - start-up slot'
    if row['ore_tonnes'] < 0.25 * row['estimated_tonnes'] or (row['average_grade'] or 0) < 0.35:
        return 'Low-grade stockpile'
    if (row['average_grade'] or 0) >= 1.10:
        return 'Crusher direct feed'
    if (row['hardness_factor'] or 1) > 1.15:
        return 'ROM blend stockpile'
    return 'ROM pad'

patterns['destination'] = patterns.apply(destination, axis=1)

# Risk flags and quality flags. Put hardness/domain flags where the prototype can find them.
def make_flags(row):
    flags = []
    if row['pattern_type'] in ['Ramp', 'Slot', 'Trim']:
        flags.append('development_or_boundary_control')
    if (row['hardness_factor'] or 1) > 1.15:
        flags.append(f'hardness_factor:{row["hardness_factor"]:.2f}')
    if str(row['digability']).lower() == 'hard':
        flags.append('hard_digability')
    if (row['predicted_p80_mm'] or 0) > 145:
        flags.append('coarse_fragmentation_risk')
    if (row['average_grade'] or 0) < 0.35:
        flags.append('low_grade_or_dilution_risk')
    if (row['wet_fraction'] or 0) > 0.25:
        flags.append('wet_material_risk')
    if str(row['primary_mineral_domain']).lower() in ['fault_damage_zone', 'late_barren_intrusion']:
        flags.append(f'domain:{row["primary_mineral_domain"]}')
    return flags

patterns['risk_flags_list'] = patterns.apply(make_flags, axis=1)
patterns['data_quality_flags_list'] = patterns.apply(lambda r: [] if r['block_count'] > 0 and not pd.isna(r['average_grade']) else ['missing_block_model_or_grade'], axis=1)
patterns['geometry_flags_list'] = patterns.apply(lambda r: ['small_or_special_pattern'] if r['area_m2'] < 2500 or r['pattern_type'] in ['Ramp', 'Slot', 'Trim'] else [], axis=1)

# Development/access priority.
type_dev_priority = {'Ramp': 100, 'Slot': 95, 'Trim': 70, 'Production': 10}
patterns['development_priority'] = patterns['pattern_type'].map(type_dev_priority).fillna(10).astype(int)
patterns['access_priority'] = patterns['centroid_y'].rank(method='dense').astype(int)

# Demo sequence. This is intentionally generated to look like an external balanced-feed sequence.
manual_sequence = [
    1, 2,
    4, 7,
    3, 9,
    5, 11,
    6, 10,
    8, 13,
    16, 20,
    14, 17,
    12, 15,
    18, 19,
    21, 22,
    23, 24,
    25, 26,
    27,
]
patterns = patterns.set_index('pattern_id').loc[manual_sequence].reset_index()
patterns['sequence_index'] = np.arange(1, len(patterns) + 1)
patterns['period_number'] = ((patterns['sequence_index'] - 1) // 2) + 1
patterns.loc[patterns['sequence_index'] == len(patterns), 'period_number'] = 14
patterns['period_id'] = patterns['period_number'].apply(lambda n: f'D{int(n):02d}')
patterns['period_label'] = patterns['period_number'].apply(lambda n: f'Day {int(n)}')

starts = []
finishes = []
for _, row in patterns.iterrows():
    day_idx = int(row['period_number']) - 1
    slot_idx = (int(row['sequence_index']) - 1) % 2
    # First slot 06:00-15:00, second 15:00-23:00. Final single slot gets a longer window.
    start = START + timedelta(days=day_idx, hours=slot_idx * 9)
    window_hours = 9 if len(patterns[patterns['period_number'] == row['period_number']]) > 1 else 12
    finish = start + timedelta(hours=window_hours)
    starts.append(start.isoformat().replace('+00:00', 'Z'))
    finishes.append(finish.isoformat().replace('+00:00', 'Z'))
patterns['scheduled_start'] = starts
patterns['scheduled_finish'] = finishes
patterns['scheduled_period'] = patterns['period_id']
patterns['source_system'] = 'Demo external scheduler'
patterns['schedule_name'] = 'Demo external two-week balanced feed sequence'
patterns['planned_loader'] = patterns['pattern_type'].map({'Ramp': 'EX-02', 'Slot': 'EX-02', 'Trim': 'EX-03', 'Production': 'EX-01'}).fillna('EX-01')
patterns['planned_truck_fleet'] = patterns['pattern_type'].map({'Ramp': 'Fleet B', 'Slot': 'Fleet B', 'Trim': 'Fleet B', 'Production': 'Fleet A'}).fillna('Fleet A')
patterns['dependency_pattern_ids'] = patterns.apply(lambda r: '' if r['sequence_index'] <= 1 else str(int(patterns.loc[patterns['sequence_index'] == r['sequence_index'] - 1, 'pattern_id'].iloc[0])), axis=1)
patterns['dependency_pattern_ids'] = patterns['dependency_pattern_ids'].fillna('').astype(str).replace({'nan': ''})
patterns['why'] = patterns.apply(lambda r: (
    'Access/development pattern scheduled early to open the bench.' if r['pattern_type'] in ['Ramp', 'Slot'] else
    'Production pattern sequenced to maintain southern-to-northern bench progression and stable mill feed.' if r['pattern_type'] == 'Production' else
    'Trim/boundary pattern sequenced late to clean up edge material and maintain next-period access.'
), axis=1)

# Create export-friendly columns.
patterns['risk_flags'] = patterns['risk_flags_list'].apply(lambda xs: ';'.join(xs))
patterns['data_quality_flags'] = patterns['data_quality_flags_list'].apply(lambda xs: ';'.join(xs))
patterns['geometry_flags'] = patterns['geometry_flags_list'].apply(lambda xs: ';'.join(xs))
patterns['contained_metal_t'] = patterns['contained_cu_t']
patterns['recovered_metal_t'] = patterns['recovered_cu_t']
patterns['metal'] = 'Cu'
patterns['pit'] = 'Site B / Pit A'
patterns['mine_area'] = 'Bench RL185 / BM v3'

csv_columns = [
    'sequence_index', 'period_id', 'period_label', 'scheduled_period', 'scheduled_start', 'scheduled_finish',
    'pattern_id', 'pattern_key', 'pattern_type', 'pit', 'mine_area', 'bench', 'phase', 'pushback',
    'source_system', 'schedule_name', 'planned_loader', 'planned_truck_fleet', 'destination',
    'estimated_tonnes', 'ore_tonnes', 'waste_tonnes', 'average_grade', 'contained_metal_t', 'recovered_metal_t', 'metal',
    'hole_count', 'drill_metres_m', 'explosive_kg', 'powder_factor_kg_m3', 'powder_factor_kg_t',
    'burden_m', 'spacing_m', 'hole_diameter_mm', 'target_p80_mm', 'predicted_p50_mm', 'predicted_p80_mm', 'predicted_p95_mm',
    'hardness_index', 'hardness_factor', 'BWi_kWht', 'UCS_MPa', 'Axb', 'DWi_kWhm3', 'digability',
    'primary_mineral_domain', 'primary_lithology', 'structure_domain', 'wet_fraction',
    'area_m2', 'perimeter_m', 'centroid_x', 'centroid_y', 'density_tpm3', 'ore_fraction',
    'haul_distance_m', 'truck_hours', 'loader_hours', 'drill_duration_hr', 'charge_duration_hr', 'load_haul_duration_hr', 'total_duration_hr',
    'development_priority', 'access_priority', 'dependency_pattern_ids', 'risk_flags', 'data_quality_flags', 'geometry_flags', 'why'
]
sequence_csv = patterns[csv_columns].copy()

# Round numeric values for a clean demo CSV.
for col in sequence_csv.select_dtypes(include=[np.number]).columns:
    if col in ['pattern_id', 'sequence_index', 'period_number', 'hole_count', 'development_priority', 'access_priority']:
        sequence_csv[col] = sequence_csv[col].round(0).astype(int)
    else:
        sequence_csv[col] = sequence_csv[col].round(3)

sequence_csv_path = OUT_DIR / 'demo_external_scheduler_sequence.csv'
sequence_csv.to_csv(sequence_csv_path, index=False)

# Period summaries.
period_rows = []
for period_id, group in patterns.groupby('period_id', sort=False):
    ore_tonnes = float(group['ore_tonnes'].sum())
    waste_tonnes = float(group['waste_tonnes'].sum())
    total_tonnes = float(group['estimated_tonnes'].sum())
    contained = float(group['contained_cu_t'].sum())
    recovered = float(group['recovered_cu_t'].sum())
    avg_grade = float((contained / ore_tonnes) * 100.0) if ore_tonnes > 0 else None
    period_rows.append({
        'period_id': period_id,
        'period_label': group['period_label'].iloc[0],
        'start': group['scheduled_start'].min(),
        'end': group['scheduled_finish'].max(),
        'pattern_ids': ','.join(group['pattern_id'].astype(int).astype(str)),
        'pattern_count': int(len(group)),
        'ore_tonnes': round(ore_tonnes, 3),
        'waste_tonnes': round(waste_tonnes, 3),
        'total_tonnes': round(total_tonnes, 3),
        'average_grade': None if avg_grade is None else round(avg_grade, 4),
        'contained_cu_t': round(contained, 3),
        'recovered_cu_t': round(recovered, 3),
        'drill_metres_m': round(float(group['drill_metres_m'].sum()), 3),
        'explosive_kg': round(float(group['explosive_kg'].sum()), 3),
        'truck_hours': round(float(group['truck_hours'].sum()), 3),
        'loader_hours': round(float(group['loader_hours'].sum()), 3),
        'strip_ratio': None if ore_tonnes == 0 else round(waste_tonnes / ore_tonnes, 3),
        'warnings': ';'.join(sorted(set(';'.join(group['risk_flags']).split(';')) - {''})),
    })
periods_df = pd.DataFrame(period_rows)
periods_csv_path = OUT_DIR / 'demo_external_scheduler_periods.csv'
periods_df.to_csv(periods_csv_path, index=False)

# Build JSON schedule asset matching the prototype schema as closely as possible.
generated_at = datetime.now(timezone.utc).isoformat().replace('+00:00', 'Z')
schedule_id = 'external-demo-sequence-A185-BM-v3'
scenario_id = 'external-balanced-feed-2w'

pattern_assets = []
for _, r in patterns.iterrows():
    pattern_assets.append({
        'pattern_id': str(int(r['pattern_id'])),
        'pattern_key': r['pattern_key'],
        'pattern_type': r['pattern_type'],
        'bench': r['bench'],
        'phase': r['phase'],
        'pushback': r['pushback'],
        'estimated_tonnes': float(r['estimated_tonnes']),
        'ore_tonnes': float(r['ore_tonnes']),
        'waste_tonnes': float(r['waste_tonnes']),
        'mineralised_tonnes': float(r['ore_tonnes']),
        'average_grade': None if pd.isna(r['average_grade']) else float(r['average_grade']),
        'contained_metal_t': None if pd.isna(r['contained_cu_t']) else float(r['contained_cu_t']),
        'recovered_metal_t': None if pd.isna(r['recovered_cu_t']) else float(r['recovered_cu_t']),
        'metal': 'Cu',
        'value_proxy': None if pd.isna(r['recovered_cu_t']) else float(r['recovered_cu_t']),
        'destination': r['destination'],
        'truck_hours': float(r['truck_hours']),
        'loader_hours': float(r['loader_hours']),
        'haul_distance_m': float(r['haul_distance_m']),
        'drill_metres_m': float(r['drill_metres_m']),
        'explosive_kg': float(r['explosive_kg']),
        'hole_count': int(r['hole_count']),
        'powder_factor_kg_m3': None if pd.isna(r['powder_factor_kg_m3']) else float(r['powder_factor_kg_m3']),
        'powder_factor_kg_t': None if pd.isna(r['powder_factor_kg_t']) else float(r['powder_factor_kg_t']),
        'predicted_p80_mm': None if pd.isna(r['predicted_p80_mm']) else float(r['predicted_p80_mm']),
        'target_p80_mm': None if pd.isna(r['target_p80_mm']) else float(r['target_p80_mm']),
        'hardness_factor': None if pd.isna(r['hardness_factor']) else float(r['hardness_factor']),
        'BWi_kWht': None if pd.isna(r['BWi_kWht']) else float(r['BWi_kWht']),
        'UCS_MPa': None if pd.isna(r['UCS_MPa']) else float(r['UCS_MPa']),
        'Axb': None if pd.isna(r['Axb']) else float(r['Axb']),
        'DWi_kWhm3': None if pd.isna(r['DWi_kWhm3']) else float(r['DWi_kWhm3']),
        'digability': r['digability'],
        'primary_mineral_domain': r['primary_mineral_domain'],
        'primary_lithology': r['primary_lithology'],
        'drill_duration_hr': float(r['drill_duration_hr']),
        'charge_duration_hr': float(r['charge_duration_hr']),
        'load_haul_duration_hr': float(r['load_haul_duration_hr']),
        'total_duration_hr': float(r['total_duration_hr']),
        'risk_flags': list(r['risk_flags_list']),
        'data_quality_flags': list(r['data_quality_flags_list']),
        'geometry_flags': list(r['geometry_flags_list']),
        'dependency_pattern_ids': [] if r['dependency_pattern_ids'] == '' else [str(r['dependency_pattern_ids'])],
        'unlocks_pattern_ids': [],
        'access_priority': int(r['access_priority']),
        'development_priority': int(r['development_priority']),
        'is_isolated': False,
        'scheduled_period': r['scheduled_period'],
        'scheduled_start': r['scheduled_start'],
        'scheduled_finish': r['scheduled_finish'],
        'sequence_index': int(r['sequence_index']),
        'source_system': r['source_system'],
        'planned_loader': r['planned_loader'],
        'planned_truck_fleet': r['planned_truck_fleet'],
        'why': [r['why']],
    })

period_assets = []
for _, r in periods_df.iterrows():
    period_assets.append({
        'period_id': r['period_id'],
        'label': r['period_label'],
        'start': r['start'],
        'end': r['end'],
        'pattern_ids': str(r['pattern_ids']).split(','),
        'ore_tonnes': float(r['ore_tonnes']),
        'waste_tonnes': float(r['waste_tonnes']),
        'total_tonnes': float(r['total_tonnes']),
        'average_grade': None if pd.isna(r['average_grade']) else float(r['average_grade']),
        'contained_cu_t': float(r['contained_cu_t']),
        'recovered_cu_t': float(r['recovered_cu_t']),
        'drill_metres_m': float(r['drill_metres_m']),
        'explosive_kg': float(r['explosive_kg']),
        'truck_hours': float(r['truck_hours']),
        'loader_hours': float(r['loader_hours']),
        'duration_hr': 24.0,
        'strip_ratio': None if pd.isna(r['strip_ratio']) else float(r['strip_ratio']),
        'warnings': [] if pd.isna(r['warnings']) or not str(r['warnings']).strip() else str(r['warnings']).split(';'),
    })

total_ore = float(patterns['ore_tonnes'].sum())
total_waste = float(patterns['waste_tonnes'].sum())
total_tonnes = float(patterns['estimated_tonnes'].sum())
total_contained = float(patterns['contained_cu_t'].sum())
total_recovered = float(patterns['recovered_cu_t'].sum())
avg_grade_total = float((total_contained / total_ore) * 100.0) if total_ore > 0 else None
strip_ratio = float(total_waste / total_ore) if total_ore > 0 else None

# Build scenario object with the field names used by the current React card where practical.
scenario_periods = []
for p in period_assets:
    pstart = datetime.fromisoformat(p['start'].replace('Z', '+00:00'))
    rows = []
    for pid in p['pattern_ids']:
        pr = next(x for x in pattern_assets if x['pattern_id'] == str(pid))
        start_dt = datetime.fromisoformat(pr['scheduled_start'].replace('Z', '+00:00'))
        finish_dt = datetime.fromisoformat(pr['scheduled_finish'].replace('Z', '+00:00'))
        rows.append({
            'patternId': pr['pattern_id'],
            'scheduledStartHr': (start_dt - START).total_seconds() / 3600.0,
            'scheduledFinishHr': (finish_dt - START).total_seconds() / 3600.0,
            'reason': pr['why'][0],
        })
    scenario_periods.append({
        'periodId': p['period_id'],
        'label': p['label'],
        'startOffsetHr': (pstart - START).total_seconds() / 3600.0,
        'endOffsetHr': (datetime.fromisoformat(p['end'].replace('Z', '+00:00')) - START).total_seconds() / 3600.0,
        'rows': rows,
        'oreTonnes': p['ore_tonnes'],
        'wasteTonnes': p['waste_tonnes'],
        'totalTonnes': p['total_tonnes'],
        'averageGrade': p['average_grade'],
        'stripRatio': p['strip_ratio'],
        'drillMetresM': p['drill_metres_m'],
        'explosiveKg': p['explosive_kg'],
        'truckHours': p['truck_hours'],
        'constraintFlags': p['warnings'],
    })

scenario_candidates = []
for p in pattern_assets:
    scenario_candidates.append({
        'patternId': p['pattern_id'],
        'patternType': p['pattern_type'],
        'bench': p['bench'],
        'phase': p['phase'],
        'pushback': p['pushback'],
        'estimatedTonnes': p['estimated_tonnes'],
        'oreTonnes': p['ore_tonnes'],
        'wasteTonnes': p['waste_tonnes'],
        'mineralisedTonnes': p['mineralised_tonnes'],
        'averageGrade': p['average_grade'],
        'valueProxy': p['value_proxy'],
        'destination': p['destination'],
        'truckHours': p['truck_hours'],
        'haulDistanceM': p['haul_distance_m'],
        'drillMetresM': p['drill_metres_m'],
        'explosiveKg': p['explosive_kg'],
        'drillDurationHr': p['drill_duration_hr'],
        'chargeDurationHr': p['charge_duration_hr'],
        'loadHaulDurationHr': p['load_haul_duration_hr'],
        'totalWorkDurationHr': p['total_duration_hr'],
        'riskFlags': p['risk_flags'],
        'dataQualityFlags': p['data_quality_flags'],
        'geometryFlags': p['geometry_flags'],
        'dependencyPatternIds': p['dependency_pattern_ids'],
        'unlocksPatternIds': p['unlocks_pattern_ids'],
        'accessRank': p['access_priority'],
        'developmentPriority': p['development_priority'],
    })

summary = {
    'pattern_count': int(len(patterns)),
    'scheduled_pattern_count': int(len(patterns)),
    'unscheduled_pattern_count': 0,
    'total_ore_tonnes': total_ore,
    'total_waste_tonnes': total_waste,
    'total_tonnes': total_tonnes,
    'total_drill_metres_m': float(patterns['drill_metres_m'].sum()),
    'total_explosive_kg': float(patterns['explosive_kg'].sum()),
    'average_grade': avg_grade_total,
    'contained_cu_t': total_contained,
    'recovered_cu_t': total_recovered,
    'strip_ratio': strip_ratio,
    'total_duration_hr': float(patterns['total_duration_hr'].sum()),
    'constraint_violation_count': int(sum(len(p['warnings']) for p in period_assets)),
}

schedule_asset = {
    'stage': 'blast_master_schedule',
    'schedule_id': schedule_id,
    'source_system': 'Demo external scheduler',
    'source_file': 'demo_external_scheduler_sequence.csv',
    'generated_at': generated_at,
    'horizon': '2_weeks',
    'period_granularity': 'day',
    'objective': 'balanced_feed',
    'constraints': {
        'start_date_iso': START.isoformat().replace('+00:00', 'Z'),
        'horizon': '2_weeks',
        'granularity': 'day',
        'shift_hours': 12,
        'shifts_per_day': 2,
        'drill_capacity_m_per_period': 2500,
        'charge_capacity_kg_per_period': 80000,
        'truck_hours_per_period': 240,
        'loader_hours_per_period': 24,
        'ore_target_tonnes_per_period': 220000,
        'waste_target_tonnes_per_period': 40000,
        'total_tonnes_target_per_period': 260000,
        'target_grade': 1.0,
        'grade_min': 0.4,
        'grade_max': 1.6,
        'max_strip_ratio': 0.35,
        'min_blast_gap_hr': 2,
        'max_patterns_per_period': 2,
        'max_blasts_per_period': 2,
        'default_density_t_per_m3': 2.7,
        'enable_pit_shell_clipping': True,
        'require_inside_pit_shell': False,
        'enable_spatial_dependencies': True,
        'enable_destination_routing': True,
        'enable_stockpile_balancing': True,
        'enable_haulage_proxy': True,
        'enable_risk_penalty': True,
    },
    'patterns': pattern_assets,
    'period_summaries': period_assets,
    'summary': summary,
    'scenario': {
        'scenarioId': scenario_id,
        'name': 'External scheduler demo - balanced feed 2 week sequence',
        'generatedAt': generated_at,
        'objective': 'balanced_feed',
        'constraints': {
            'horizon': '2_weeks',
            'granularity': 'day',
            'shiftHours': 12,
            'shiftsPerDay': 2,
            'targetOreTonnesPerPeriod': 220000,
            'targetWasteTonnesPerPeriod': 40000,
            'targetTotalTonnesPerPeriod': 260000,
            'targetGrade': 1.0,
            'minGrade': 0.4,
            'maxGrade': 1.6,
            'maxStripRatio': 0.35,
            'maxPatternsPerPeriod': 2,
            'minBlastGapHr': 2,
            'maxBlastsPerPeriod': 2,
            'enablePitShellClipping': True,
            'requireInsidePitShell': False,
            'enableSpatialDependencies': True,
            'enableDestinationRouting': True,
            'enableStockpileBalancing': True,
            'enableHaulageProxy': True,
            'enableRiskPenalty': True,
        },
        'candidates': scenario_candidates,
        'periods': scenario_periods,
        'summary': {
            'candidateCount': int(len(patterns)),
            'scheduledPatternCount': int(len(patterns)),
            'unscheduledPatternCount': 0,
            'totalTonnes': total_tonnes,
            'totalOreTonnes': total_ore,
            'totalWasteTonnes': total_waste,
            'totalDrillMetresM': float(patterns['drill_metres_m'].sum()),
            'totalExplosiveKg': float(patterns['explosive_kg'].sum()),
            'totalTruckHours': float(patterns['truck_hours'].sum()),
            'averageGrade': avg_grade_total,
            'stripRatio': strip_ratio,
            'constraintViolationCount': int(sum(len(p['warnings']) for p in period_assets)),
            'highRiskPatternCount': int(sum(1 for p in pattern_assets if p['risk_flags'])),
            'targetAchievement': {
                'totalTonnesPct': None,
                'oreTonnesPct': None,
                'gradeVariance': None if avg_grade_total is None else avg_grade_total - 1.0,
            },
        },
    },
}

# Convert NaN to null recursively.
def clean_json(obj):
    if isinstance(obj, dict):
        return {k: clean_json(v) for k, v in obj.items()}
    if isinstance(obj, list):
        return [clean_json(v) for v in obj]
    if isinstance(obj, (np.integer,)):
        return int(obj)
    if isinstance(obj, (np.floating,)):
        obj = float(obj)
    if isinstance(obj, float) and (math.isnan(obj) or math.isinf(obj)):
        return None
    return obj

schedule_asset = clean_json(schedule_asset)
json_path = OUT_DIR / 'demo_external_scheduler_schedule_asset.json'
json_path.write_text(json.dumps(schedule_asset, indent=2), encoding='utf-8')

# README
readme = f'''# Demo external scheduler sequence package

This package contains a synthetic external scheduler sequence generated from the supplied demo data:

- `blast_master.csv`
- `block_model_copper.csv`
- `drill_charge_designs.zip`

It is not an output from a commercial scheduling package. It is an import-ready demo sequence designed to exercise the Scenario Simulation workflow.

## Files

- `demo_external_scheduler_sequence.csv` — pattern-level sequence for import into the Scenario Simulation workspace.
- `demo_external_scheduler_periods.csv` — day-level period summaries.
- `demo_external_scheduler_schedule_asset.json` — full schedule asset matching the current Blast Master prototype schema as closely as possible.
- `create_demo_external_sequence.ipynb` — notebook to regenerate or tune the sequence from the same source files.

## Sequence basis

- Horizon: 2 weeks / 14 daily periods.
- Bench: RL185-200.
- Patterns: {len(patterns)} total.
- Scheduling logic: Ramp and slot first, production patterns next, trim/boundary patterns late.
- Objective: balanced feed and bench progression for demonstration.
- Start date: {START.isoformat().replace('+00:00', 'Z')}.

## Demo summary

- Total scheduled tonnes: {total_tonnes:,.0f} t
- Ore tonnes: {total_ore:,.0f} t
- Waste tonnes: {total_waste:,.0f} t
- Average Cu grade: {avg_grade_total:.3f}%
- Contained Cu: {total_contained:,.0f} t
- Recovered Cu proxy: {total_recovered:,.0f} t
- Drill metres: {float(patterns['drill_metres_m'].sum()):,.0f} m
- Explosive mass: {float(patterns['explosive_kg'].sum()):,.0f} kg

## Recommended use

Use `demo_external_scheduler_sequence.csv` if the workspace import expects a flat external scheduler file. Use `demo_external_scheduler_schedule_asset.json` if the import expects the local Blast Master schedule asset shape.

If the importer requires different column names, map these core columns:

- `pattern_id`
- `sequence_index`
- `scheduled_period`
- `scheduled_start`
- `scheduled_finish`
- `destination`
- `estimated_tonnes`
- `ore_tonnes`
- `waste_tonnes`
- `average_grade`
- `drill_metres_m`
- `explosive_kg`
- `risk_flags`
'''
readme_path = OUT_DIR / 'README_demo_external_sequence.md'
readme_path.write_text(readme, encoding='utf-8')

print('Created:')
for p in [sequence_csv_path, periods_csv_path, json_path, readme_path]:
    print('-', p)
```
