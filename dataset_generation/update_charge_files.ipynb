# Insert deck fields into charge files

```python
from pathlib import Path
import pandas as pd

# =========================
# USER INPUTS
# =========================

# Add a variable that is set to the parent folder
input_folder = Path(r"data/charge_actuals")
output_folder = input_folder / "updated_with_decks"

default_deck = 1
default_deck_type = "charge"

# Set to True if you want to overwrite the original files instead
overwrite_originals = False

# =========================
# PROCESS FILES
# =========================

output_folder.mkdir(parents=True, exist_ok=True)

csv_files = sorted(input_folder.glob("*.csv"))

if not csv_files:
    raise FileNotFoundError(f"No CSV files found in: {input_folder}")

for file_path in csv_files:
    df = pd.read_csv(file_path)

    # Add deck fields if they do not already exist
    if "deck" not in df.columns:
        df["deck"] = default_deck

    if "deck_type" not in df.columns:
        df["deck_type"] = default_deck_type

    # Place deck fields near the charging/product fields if possible
    preferred_order = []

    # Keep these leading identifiers first if present
    leading_cols = [
        "pattern_number",
        "hole_number",
        "hole_id",
    ]

    for col in leading_cols:
        if col in df.columns:
            preferred_order.append(col)

    # Insert deck fields immediately after hole identifiers
    for col in ["deck", "deck_type"]:
        if col in df.columns and col not in preferred_order:
            preferred_order.append(col)

    # Then keep all remaining columns in their original relative order
    remaining_cols = [col for col in df.columns if col not in preferred_order]
    df = df[preferred_order + remaining_cols]

    # Decide output path
    if overwrite_originals:
        output_path = file_path
    else:
        output_path = output_folder / file_path.name

    df.to_csv(output_path, index=False)

    print(f"Updated: {file_path.name} -> {output_path}")

print(f"\nDone. Processed {len(csv_files)} file(s).")
```
